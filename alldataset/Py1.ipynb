```python
import numpy as np
import matplotlib.pyplot as plt

# Create x values
x = np.linspace(0.1, 10, 500)  # Avoid 0 to prevent log(0) and tan(π/2) issues

# Compute function values
y_const = 3
y_log = np.log(x)
y_lin = x
y_xlog = x*np.log(x)
y_xx = x*x
y_exp = np.exp(x)

# Plot each function with a different color
#plt.plot(x, y_const, label='const', color='blue')
plt.plot(x, y_log, label='log', color='green')
plt.plot(x, y_lin, label='linear', color='red')
plt.plot(x, y_xlog, label='logx', color='purple')
plt.plot(x, y_exp, label='exp(x)', color='orange')
plt.plot(x, y_xx, label='x-x', color='black')

# Add labels and legend
plt.title('Plot of sin, cos, tan, log, exp')
plt.xlabel('x')
plt.ylabel('Function values')
plt.ylim(-1, 10)  # Limit y-axis to avoid tan(x) spikes
plt.grid(True)
plt.legend()

# Show the plot
plt.show()
```
