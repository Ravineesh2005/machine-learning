```python
import pandas as pd 
import requests
from bs4 import BeautifulSoup
import numpy as np
import lxml
```

```python
headers={'User-Agent':'Mozilla/5.0 (Windows NT 6.3; Win 64 ; x64) AppleWeKit /537.36(KHTML , like Gecko) Chrome/80.0.3987.162 Safari/537.36'} 
```

```python
webpage = requests.get('https://www.ambitionbox.com/list-of-companies?page%252', headers = headers).text
```

```python
soup = BeautifulSoup(webpage,"lxml")
```

```python
company=soup.find_all('div',class_='companyCardWrapper')
len(company)
```

```python

name=[]
rating=[]
reviews=[]
salery=[]
interview=[]
job_ava=[]
location=[]

for i in company:

  name.append(i.find('h2').text.strip())
  rating.append(i.find('div',class_='rating_text').text.strip())
  location.append(i.find_all('span' , class_='companyCardWrapper__ActionCount')[0].text.strip())
  reviews.append(i.find_all('span' , class_='companyCardWrapper__ActionCount')[0].text.strip())
  salery.append(i.find_all('span' , class_='companyCardWrapper__ActionCount')[1].text.strip())
  interview.append(i.find_all('span' , class_='companyCardWrapper__ActionCount')[2].text.strip())
  job_ava.append(i.find_all('span' , class_='companyCardWrapper__ActionCount')[3].text.strip())
  #no_of_employee.append(i.find_all('p',class_='infoEntity')[3].text.strip())

```

```python
df=pd.DataFrame({'name':name,
   'rating':rating,
   'location':location,
   'reviews':reviews,
   'salery':salery,
   'interview':interview,
   'job_ava':job_ava,
   })
```
